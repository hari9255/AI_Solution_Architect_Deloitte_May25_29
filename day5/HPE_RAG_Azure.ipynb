```python
from langchain_community.document_loaders import PyPDFLoader
loader = PyPDFLoader("/content/1706.03762v7.pdf")
pages = loader.load_and_split()
len(pages)
```

```python
"""#chunking"""
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=300,
    chunk_overlap=80,

)

chunks = text_splitter.split_documents(pages)
```

```python
vector_store_address = "https://azureaisearch1705-demo.search.windows.net"
vector_store_password = "jsL51d8I4PS7RqKE37keWrk2qARIMmknD9ZtTnd81cAzSeCFUwOk"
index_name = "agilefever-rag-demo-harish-1111"
```

```python
vector_store = AzureSearch(
    azure_search_endpoint=vector_store_address,
    index_name=index_name,
    embedding_function=embeddings.embed_query,
    azure_search_key=vector_store_password
)
```

```python
vector_store.add_documents(documents=chunks)
```

```python
user_input = input("Enter your question: ")
```

```python
relevant_chunks = vector_store.similarity_search(
    query = user_input,
    k = 2,
    search_type = "similarity",

)
```
