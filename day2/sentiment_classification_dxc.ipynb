```python
import pandas as pd
data = pd.read_csv("/content/amazon_cells_labelled (1).txt",delimiter= "\t",header=None)
```

```python
data.columns = ['feedback','sentiment']
```

```python
data.head()
```

```python
    import nltk
    from nltk.corpus import stopwords
    from nltk.tokenize import word_tokenize

    # Download the 'stopwords' and 'punkt' (for word_tokenize) resources if you haven't already
    nltk.download('stopwords')
    nltk.download('punkt')
```

```python
stop_words = set(stopwords.words('english')) # Get English stopwords as a set for efficient lookup
```

```python
stop_words
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
def remove_stopwords(text):
    # Tokenize the text into words
    words = word_tokenize(text)
    # Remove stopwords from the token
    filtered_sentence = [w for w in words if not w.lower() in stop_words]
    return ' '.join(filtered_sentence)
```

```python
data['feedback']= data['feedback'].apply(lambda x: remove_stopwords(x))
```

```python
data['feedback'][0]
```

```python
data['sentiment'][0]
```

```python
def text_preprocessing(feedback):
  return feedback.lower().replace("[^\w\s]","").replace('\d',"") # lower case, remove punctuation, remove number, remove stopwords

```

```python
data['feedback'] = data['feedback'].apply(lambda x: text_preprocessing(x))
```

```python
data['feedback'][0]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(data['feedback'],data['sentiment'],train_size=0.80,random_state=40 )
```

```python
X_train.head()
```

```python
y_train.head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer(ngram_range=(1,2),max_df=0.7,min_df=0.10,max_features=50000)
X_train = vectorizer.fit_transform(X_train)
X_test = vectorizer.transform(X_test)
```

```python
X_train
```

```python
from sklearn.linear_model import LogisticRegression
LR = LogisticRegression().fit(X_train,y_train)
```

```python
y_test_pred = LR.predict(X_test)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_test_pred))
```
