```python
reteriver = recursive_db.as_retriever(search_type ="mmr")
result = reteriver.invoke(user_query )
```

```python
result
```

```python
#improving area-1: re ranking of context

query_chunk_pair = [(user_query,chunk.page_content)for chunk in result]
```

```python
from sentence_transformers import CrossEncoder
model = CrossEncoder('cross-encoder/ms-marco-MiniLM-L-6-v2', max_length=512)
scores = model.predict(query_chunk_pair)
```

```python
content_list=[]
for relevant_chunk in result:
    content_list.append(relevant_chunk.page_content)
pairs = list(zip(scores, content_list))
pairs.sort(key=lambda pair: pair[0], reverse=True)
sorted_content = [pair[1] for pair in pairs]
```
